```python
age = int(input("age: "))

if age < 18:
    print("Children")
elif age < 60:
    print("Adult")
elif age < 80:
    print("Citizen")
else:
    print("Senior Citizen")
```

```python
num = float(input("Enter any number: "))

if num > 0:
    print(f"{num:.0f} is positive")
elif num < 0:
    print(f"{num:.0f} is negative")
else:
    print("Zero")
```

```python
num = int(input("Enter a number to check: "))

if num % 5 == 0:
    print(f"{num} is divisible by 5")
else:
    print(f"{num} is not divisible by 5")
```
